## 条件判断
本章节内容主要介绍条件判断语句。他往往用于场景中有多种可能的情况，每一种情况都有其各自的操作。

内容比较简单，只有`if-else`, `if-elif-else`这两种，其中`else`不是必要的，可以没有，但是`if`是必须的。也就是说，你想使用条件分支，必须是以`if`开始的。

**需要注意的是**：
- 每个条件后面需要使用冒号`:`，表示接下来是满足条件后需要执行的代码
- 使用缩进来划分代码块，同一级缩进（拥有相同空格数的缩进）表示同一个代码块。
- `if`和`elif`后面都需要跟一个`条件表达式`。
    - 例如`1==2`, `4 > 3`,  `"" is None`诸如此类，其结果是`True`或者`False`的表达式。
    - 通过条件表达式，你可以控制满足条件时，该做什么；不满足时，又该做什么。

关于缩进和代码块，你可以理解成是word里面的一级标题，二级标题等。例如：

章节一：<br>
&emsp;&emsp;1.1 第一小节<br>
&emsp;&emsp;1.2 第二小节<br>
章节二：<br>
&emsp;&emsp;2.1 第一小节<br>
显然，1.1和1.2隶属于章节一。而2.1隶属于章节二。<br>
python中的缩进与代码块也是同理。

**NOTE**：对于没有缩进的代码（也就是，我们从一个空白文件开始写代码时，是不需要使用缩进的），我们称他为全局代码。
##### 样例：

```python
if True:
    print("goto if")
else:
    print("goto else")

print("break if-else")
'''
根据上面的例子，很容易理解:
- print("goto if")是隶属于if的代码
- print("goto else")是隶属于else的代码
- print("break if-else")则是全局代码
'''

## if-elif-else
## 当我们的代码逻辑有不止两种条件分支时，可以使用这种形式
## 其实逻辑就和自然语言，如果...那么类似
## 例如：成绩判断，60以下是不及格，60-80良好，80-90优秀，90-100天才
## 如果 成绩 < 60: 那么输出不及格
## 反之如果成绩 < 70：那么xxx
## 反之如果成绩 < 80：那么xxx
## 反之如果成绩 < 90：那么xxx
## 反之：那么xxx //这里不需要再用如果，原因是我们已经把可能的情况列完了，只剩下最后一种情况了。
## 当前面情况都不满足时，就会自动跳到这种情况来。因此有没有条件表达式都不影响正确性。
score = 20
if score < 60:
    print("不及格")
elif 70 > score >= 60:
    print("良好")
elif 80 > score >= 70:
    print("优秀")
else:
    print("天才")

## else不是必须的
## 例如，我们只想统计班级中，成绩不及格的人数，那么我们只需要对成绩小于60的进行统计即可
scores = [2, 3, 4, 0, 100, 80, 90]
num = 0
for score in scores:
    if score < 60:
        num += 1
print(f"不及格人数：{num}")
```

##### 题目描述：
- 假设你要监控 A,B,C 三个商品
- 如果 A/B/C 分别降价 30/20/10 元上，就要报警
- 如果需要报警则打印商品名 + 降价值

```python
import time
a_price, b_price, c_price = 80, 60, 40

while True:
    a_discount = random.randint(10, 50)
    b_discount = random.randint(5, 40)
    c_discount = random.randint(5, 25)
    a_price -= a_discount
    b_price -= b_discount
    c_price -= c_discount
    print(f"A/B/C的折扣幅度为{a_discount}/{b_discount}/{c_discount}")

    # a_discount, b_discount, c_discount分别是A, B, C的降价幅度
    # 请根据题目要求补全代码
    # 可以根据上一题那样去思考。一般地，coding逻辑就是这样

    if a_price <= 0 or b_price <= 0 or c_price <= 0:
        break
    time.sleep(1)
```
