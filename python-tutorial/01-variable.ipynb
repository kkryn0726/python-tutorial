#### 习题3：输入与输出
在python中，输入与输出分别是`input`和`print`函数，用于获取从键盘输入的数据，和打印内容到控制台。

`input`输入函数，以回车作为结束符。也就是说，它会获取用户输入的一行数据，当用户输入回车时，结束输入。
**NOTE**：`input`的输入默认是字符串类型，如果你输入的是数字，你需要显示的使用`int()`、`float()`等方式将其转化。

`print`输出函数，会将用户传入的数据，依次打印在控制台。这里，个人建议记住一种字符串format方法即可，下面会介绍，简单快捷。

````
name = input("输入你的名字：") #在input中，你可以添加字符串，用作输入提示
age = input("输入你的年龄：") #此时你的年龄仍然是str类型，需要类型转换
# 可以使用type查看类型
print(type(age))
age = int(age) # 也可以直接在input时添加类型转换：int(input("输入你的年龄："))

'''
下面介绍一种简单的字符串format
例如接下来你想要输出：你叫xxx，今年xx岁。
可以通过该方式构建字符串：f"你叫{}，今年{}岁。"
{}表示占位符。在{}里面，你需要填写变量名，表示这个地方你想要输出的变量内容是什么。
note：字符串的引号前，需要有一个f来告诉python，我正在使用字符串format，{}是占位符，而不是简单字符。
'''

# 可以直接在print输出
print(f"你叫{name}，今年{age}岁。")
# 也可以赋值给一个字符串变量，然后print打印该变量
s = f"你叫{name}，今年{age}岁。"
print(s) # 两者效果是一样的
# 字符串format基本只需要会这种就可以，是最简便的。
# 至于其他的格式控制，可以先不用管，使用场景并不多

# 如果你有学到下面这种方式的format，请忽略，那是python2的，现在基本不使用了
print("%s" % s)
````

#### 习题四：变量赋值&运算符

变量赋值，很简单，就是单变量赋值和多变量赋值。例如：
```python
a = 1
b = 2
```
或者
```python
a, b = 1, 2
```

运算符，主要就是注意优先级问题。最简单的，例如，加减乘除，先算乘除再算加减；有括号，先算括号内的。在python中也是一样的。
**NOTE**，用于判断等于的运算符是`==`，这里一定要记住，不要和赋值用的`=`搞混。其他的数值比较就是`!=`, `>`, `>=`, `<`, `<=`...

再者，比较重要的运算符就是**逻辑运算符**，有与(`and`)、或(`or`)、非(`not`)。

- `and`，逻辑与，他要求`and`左右的事件同时为`True`，其结果才是`True`。
- `or`，逻辑或，他要求`or`左右事件有一个为`True`时，其结果就是`True`。
- `not`，逻辑非，也就是取反。这个很简单，`True`取反就是`False`，`False`取反就是`True`。

顺带提一下，`True`和`False`表示`bool`类型。其本质也是整型数值，`False`为`0`；其余不为`0`的，我们都可以认为是`True`。特别地，例如**空字符串（`""`）**，**空列表（`[]`）**等空对象，也认为是`False`。

````
a = 1
b = 2
c, d = 3, 4
print(a, b, c, d)

# 逻辑运算
x = (1 == 1) #当你不确定运算优先级的时候，可以手动添加括号，来保证运算顺序是你想要的
y = (2 == 2)
z = (3 == 4)
w = (2 == 5)
print(f"logic of x is {x}, y is {y}, z is {z}")
# 这里使用上面介绍的format就很方便
# 而且，{}不仅可以是变量名，也可以在里面进行运算操作等
print(f"x and y = {x and y}, x and z = {x and z}")
print(f"y or z = {y or z}, z or w = {z or w}")
print(f"not x = {not x}, not z = {not z}")
````

#### 习题五：类型转换
一般会用到的类型转换有：
- 整型转浮点型
- 浮点型转整型
- 字符串转整型
- 字符串转浮点型
- 数字转字符串

````
x, y = 3.1415, 1
print(int(x), float(y)) # x只取整数部分，不会四舍五入

s = "3.1415"
print(float(s), type(float(s)))
print(int(s)) # 这里要注意，因为字符串s的内容是一个浮点型数字，使用int转换，会报错
````
